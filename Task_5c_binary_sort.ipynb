```python
def binary_search(sorted_buildings: list[tuple[str, str]], target_code: str) -> int:
    """ Searches for target_code in a sorted list of (code, name) tuples. Returns the index if found, or -1 if not found. """
    low = 0
    high = len(sorted_buildings) - 1
    target_code = target_code.upper()

    while low <= high:
        mid = (low + high) // 2
        mid_code = sorted_buildings[mid][0]

        if mid_code == target_code:
            return mid # Target building found
        elif mid_code < target_code:
            low = mid + 1 # Search the right half
        else:
            high = mid - 1 # Search the left half

    return -1        
```

```python
if __name__ == "__main__":
    # Sorted alphabetically by building code
    buildings = [
        ("ADM", "Administration"),
        ("CAF", "Cafeteria"),
        ("HST", "Hostel"),
        ("ICT", "ICT lab"),
        ("LEC1", "Lecture hall 1"),
        ("LEC2", "Lecture hall 2"),
        ("LIB", "Library"),
        ("MED", "Sickbay"),
        ("RAD", "Campus radio"),
        ("SPT", "Sports office")
    ]

    search_code = "LIB"
    index = binary_search(buildings, search_code)

    if index != -1:
        print(f"Found {buildings[index][0]} ({buildings[index][1]}) at index {index}.")
    else:
        print(f"Building {search_code} not found.")
```

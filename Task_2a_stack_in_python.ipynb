```python
class NavigationStack:
    def __init__(self):
        self.stack = []

    def push_location(self, building_name):
        self.stack.append(building_name)

    def pop_location(self):
        if not self.is_empty():
            return self.stack.pop()
        return "No history to undo"

    def peek_current_location(self):
        if not self.is_empty():
            return self.stack[-1]
        return "History is empty"

    def is_empty(self):
        return len(self.stack) == 0
```

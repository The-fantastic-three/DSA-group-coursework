```python
class LinearProbingHashTable:
    """Hash table of fixed size using h(k) = k % size and linear probing."""
    def __init__(self, size: int = 11):
        self.size = size
        self.table = [None] * size

    def _hash(self, key: int) -> int:
        return key % self.size

    def insert(self, key: int) -> int:
        """Inserts a key using linear probing and returns the number of probes used."""
        start_index = self._hash(key)
        probes = 0

        for i in range(self.size):
            probes += 1
            index = (start_index + i) % self.size
            if self.table[index] is None:
                self.table[index] = key
                return probes

        raise OverflowError("Hash table is full!")

    def search(self, key: int) -> int:
        """Returns the index of the key if found, or -1 if not found."""
        start_index = self._hash(key)
        for i in range(self.size):
            index = (start_index + i) % self.size
            if self.table[index] is None:
                return -1
            if self.table[index] == key:
                return index
        return -1

    def display(self) -> None:
        for idx, val in enumerate(self.table):
            print(f"Slot {idx:2d}: {val}")
```

```python
if __name__ == "__main__":
    ht = LinearProbingHashTable(size=11)
    keys =[22, 41, 53, 46, 30, 13, 1, 67]

    total_probes = 0
    for k in keys:
        p = ht.insert(k)
        total_probes += p
        print(f"Inserted {k:2d} - probes used: {p}")

    print("\nFinal Hash Table:")
    ht.display()
    print(f"\nTotal Probes (Linear Probing): {total_probes}")
```

```python
class BookingQueue:
    def __init__(self):
        self.queue = []

    def enqueue_request(self, booking_details):
        self.queue.append(booking_details)

    def  dequeue_request(self):
        if not self.is_empty():
            return self.queue.pop(0)
        return "No booking requests to process"

    def peek_next(self):
        if not self.is_empty():
            return self.queue[0]
        return "Queue is empty"

    def is_empty(self):
        return len(self.queue) == 0
```
